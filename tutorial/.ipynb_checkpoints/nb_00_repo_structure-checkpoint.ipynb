```python
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle

def plot_repo_structure():
    fig, ax = plt.subplots(figsize=(10, 6))
    ax.set_xlim(0, 8)
    ax.set_ylim(-0.1, 6)
    ax.axis("off")  # Nasconde gli assi

    # Blocchi dei file con descrizioni
    files = [
        ("_stream.py", "Main class to read, process, visualize, transform seismic shot gathers"),
        ("_curve.py", "Class to read, process, visualize, save dispersion curves"),
        ("_combineCurves.py", "Class to combine multiple dispersion curves for error estimation"),
        ("utils.py", "Utilities, mostly for file handling"),
        ("interactive_tools.py", "Interactive tools for FK filtering, DC picking, velocity estimation, and linear muting"),
        ("geometry.py", "Functions to create and read the geometry file")
    ]
    colors = ["lightcoral", "lightyellow", "yellowgreen", "lightblue", "sandybrown", "lightgray"]
    y_positions = [5, 4, 3, 2, 1, 0]  # Posizioni verticali per ogni file

    for i, ((file, description), color) in enumerate(zip(files, colors)):
        ax.add_patch(Rectangle((1, y_positions[i]), 2, 1, color=color, ec="black", lw=1))
        ax.text(2, y_positions[i] + 0.5, file, color="black", ha="center", va="center", fontsize=11)
        ax.text(3.5, y_positions[i] + 0.5, description, color="black", ha="left", va="center", fontsize=10)

    #plt.title("Repository structure", fontsize=12, fontweight="bold")
    fig.savefig('../figures/swa_structure.png', dpi=200, transparent=True)
    plt.show()

plot_repo_structure()
#
```
