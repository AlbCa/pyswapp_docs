```python
# import packages
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from obspy import read
from scipy import signal
# import functions
import SWAutils as swa

# set directories
datadir = '../data/Synthetic_data/'
figdir = '../figures/'

#filename = input("Enter the filename and extension (.dat, .sg2, .segy){}:") 
filename = 'Shot144_coord.sg2'
```

### 1. Load data and display

```python
# load file
st = read(datadir + filename) # read the SEG-2 file with Obspy
data = np.array([trace.data for trace in st]) # extract the data from all the traces in the file
data = data.T # transpose the data array so that the first axis represents time
print(data.shape)

# plot
extent = [0, 1, 0, 1]
fig, ax = plt.subplots(1, 1, figsize=(8, 5))
cax = ax.imshow(swa.normit(data), cmap='RdBu', extent=extent)
plt.show()
```

```python
# create an instance to flip the shot (es: left/right) # user input !!!!!!!!!!!

#if 'right' in filename:
#    data = np.fliplr(data)
data = np.fliplr(data)
```

```python
# wiggle seismogram
fig = plt.figure(figsize=(10,5))
plt.title("shot gather: " +filename)
swa.wiggle(swa.normit(data))
plt.ylabel('time (trace number)')
plt.xlabel('trace')
plt.ylim(3500, 0)
```

### 2. Define acquisition parameters and select traces

```python
# acquisition scheme
x = np.arange(0,143*1, 1) # receiver # user input !!!!!!!!!!!

# sampling rate (sec)
SR = 0.00025
```

```python
# filter traces
#columns_to_keep = [col_idx for col_idx in range(data1.shape[1]) if col_idx not in [1, 5, 7]]
#data1 = data1[:, columns_to_keep] # select desired traces

#fig = plt.figure(figsize=(10,5))
#plt.title("shot gather: " +filename)
#wiggle(data1)
```

```python
# view the frequency content of a seismogram (https://community.sw.siemens.com/s/article/what-is-a-power-spectral-density-psd)
fig, ax = plt.subplots(figsize=(4.5,3.5)) 
swa.cumul_spectra(data, SR=SR, ax=ax)
ax.set_title(filename[:-4])
#fig.savefig(figdir + filename[:-4] + "_pwr_spectra.png", bbox_inches='tight', dpi=100, transparent=True)
```

### 3. Compute f-k

```python
# compute f-k and store parameters
ks, ks_w, frequencies, TRRabs_norm, TRRabs = swa.compute_fk(data, SR, x)
```

```python
# plot it
fig, ax = plt.subplots(figsize=(4.5,3.5))
swa.plot_fk(TRRabs, ks_w, frequencies, ax=ax)
fig.tight_layout()
```

#### 3.1 Customize the plot (manually)

```python
import matplotlib.colors as colors
import matplotlib.cbook as cbook
from matplotlib import cm

color_range = colors.Normalize(vmin=.1E-6, vmax=.2E-3)

fig = plt.figure(figsize=(6, 5))
plt.imshow(TRRabs, extent=[0, max(ks_w), 0, max(frequencies)], aspect='auto',# origin='lower', 
           cmap='jet',
           norm = color_range
          )
plt.xlabel('k [rad/m]')
plt.ylabel('frequency [Hz]')
plt.title('f-k spectrum: '+ filename)
plt.colorbar(shrink=0.6)
plt.ylim([5, 55])
plt.xlim([0, max(ks_w)-2.2])
fig.tight_layout()
plt.show()

#fig.savefig(figdir + filename +'fk25.png', bbox_inches='tight', transparent=True, dpi=200)
```

### 4. Perform the manual picking of max amplitudes
    intructions:
- click on the max
- when finished, click on your mouse roll
- don't close the popup manually
- run directly section 4.1 with %matplotlib inline

```python
# still need to create a function for this to be stored in SWAutils.py !!!!!!!

import matplotlib
matplotlib.use('TkAgg')

fig = plt.figure(figsize=(12,8))
plt.imshow(TRRabs, extent=[0, max(ks_w), 0, max(frequencies)], aspect='auto', #origin='lower', 
           norm = color_range,cmap='jet')
plt.xlabel('k [rad/m]')
plt.ylabel('frequency [Hz]')
plt.title('f-k spectrum')
plt.colorbar()
plt.ylim([5, 55])
plt.xlim([0, max(ks_w)-2.2])

# call 'ginput' function to interactive mark on a mpl plot
print("Click on the max amplitudes {}:")
c_max = plt.ginput(-1, timeout=0)
print(c_max)

#help(plt.ginput)
```

```python
# Extract the x and y from the 'c_max' tuple
ks_fund, freq_fund = zip(*c_max) #freq_fund, ks_fund

# resample the picked maxima
x = [p[0] for p in c_max]
y = [p[1] for p in c_max]

n = 50 # samples length
ks = np.linspace(min(x), max(x), n)
ks = np.array(ks) # new ks

freq = np.interp(ks, x, y) # interpolation
freq = np.array(freq) # new freq

# phase velocity
c = 2 * np.pi * freq / ks
```

#### 4.1 Plot the f-k and the respective dispersion curve

```python
%matplotlib inline

fig, (ax1, ax2) = plt.subplots(nrows=1, ncols=2, figsize=(11, 4))

# First subplot
im1 = ax1.imshow(TRRabs, extent=[0, max(ks_w), 0, max(frequencies)], aspect='auto',
                 norm=color_range, cmap='viridis')
ax1.plot(ks, freq, "r.", label='mode 0', markersize=3)
ax1.set_xlabel('k [rad/m]')
ax1.set_ylabel('frequency [Hz]')
ax1.set_title('f-k spectrum: ' + filename)
ax1.set_xlim([0, max(ks_w)-2])
ax1.set_ylim([0, 60])
ax1.legend()
fig.colorbar(im1, ax=ax1, shrink=0.6)

# Second subplot
ax2.scatter(freq, c, marker='.', label='fundamental mode')
ax2.plot(freq, c, linewidth=1)
ax2.set_xlabel('frequency [Hz]')
ax2.set_ylabel('phase velocity [m/s]')
ax2.set_title('dispersion curve: ' + filename)
#ax2.set_ylim([50, 140])
#ax2.set_ylim([100, 260])
ax2.legend()

# Adjust position and size of the second subplot
fig.subplots_adjust(wspace=.01)
pos = ax2.get_position()
pos.x0 = 0.6
pos.x1 = 0.8
pos.y0 = 0.14
pos.y1 = 0.6
ax2.set_position(pos)

plt.show()
fig.savefig(figdir + filename +'.png', bbox_inches='tight', transparent=True, dpi=150)
```

### 5. Save output

```python
outdir = os.path.join(datadir, 'disp_curves/')
os.makedirs(outdir, exist_ok=True) 
print(f'📁 Dispersion curve will be saved in: {outdir}')

# Save the output file as frequency-velocity
#period = 1/freq
#slowness = 1/c
out = np.column_stack((freq, c))
df = pd.DataFrame(out, columns=['freq(Hz)', 'V_ph(m/s)'])
df.to_csv(outdir+filename+'.txt', sep='\t', index=False)
df
```
